第二种：

```python
def imshow(**images):
    """
    使用字典传入窗口名和图像
    使用示例: imshow(name1=image1,name2=image2)
    name1为关键字,不是字符串
    """
    for name, image in images.items():
        cv2.imshow(name, image)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```

### 图像基本属性

#### 图像深度

图像深度（也称为位深度）指的是每个像素用于表示颜色信息的位数。图像深度决定了图像中每个像素可以表示的颜色数量，进而影响图像的质量和文件大小

#### 颜色

一张图片有BGR三个颜色通道，以灰度图的形式存储在图像矩阵的第三个维度中

- Blue image[:,:,0]
- Green image[:,:,1]
- Red image[:,:,2]

````python
image2=cv2.imread("./images/opencv_logo.jpg")
print(image2.shape) # (像素横行，像素纵列，颜色通道)
imshow(B=image2[:,:,0],G=image2[:,:,1],R=image2[:,:,2])
````

### 图像操作

图像本质上是一个多维数组，可以使用NumPy对图像进行各种操作，例如裁剪，拼接等

````python
black=np.zeros((256,256))
imshow(black=black)
````

#### 灰度图转换

对BGR通道图像做平方和加权平均

灰度图描述了图案的明暗分布，CMOS采集的图像的灰度图也可以认为是CMOS传感器接收光子数的分布图

````python
image=cv2.imread("./images/lena.png")
gray=cv2.cvtColor(image,cv2.COLOR_BGR2GRAY)
print(gray.shape)
imshow(gray=gray)
````

#### 裁剪图像

使用索引号取出图像一部分(索引号从左上角算起)

也可以使用`cv2.resize()`函数调整图像大小

````python
crop=image[50:200,:] # 第一个参数对应行，第二个对应列
imshow(crop=crop)
````

#### 图像拼接

使用`np.hstack()`和`np.vstack()`函数水平和垂直拼接图像

条件：拼接的图像尺寸必须相同

如果尺寸不同，可以使用`cv2.resize()`函数调整图像大小

````python
opening=cv2.imread("./images/opening.png")
closing=cv2.imread("./images/closing.png")

(ox,oy,_)=opening.shape
(cx,cy,_)=closing.shape
closing_resized = cv2.resize(closing, (oy, ox))
vstack_img=np.vstack((image,image))
hstack_img=np.hstack((opening,closing_resized))

imshow(hstack_img=hstack_img,vstack_img=vstack_img)
````

### 图案绘制

使用numpy生成空数组，依此生成空画布

````python
import numpy as np

image=np.zeros([300,300,3],dtype=np.uint8) # 生成黑色画布
cv2.line(image,(0,250),(300,250),(255,0,0),5) # 画线： 画布，起点坐标，终点坐标，颜色,粗细
cv2.rectangle(image,(50,50),(100,100),(0,255,0),10)# 画矩形：画布，起点坐标，终点坐标
cv2.circle(image,(200,100),50,(0,0,255),20) # 画圆：画布，圆心坐标，半径，颜色，粗细
# 厚度参数填-1代表以指定参数填充圆
cv2.putText(image,"Hello",(100,50),0,1,(255,255,255),2) # 写字：画布，内容，位置，字体序号，缩放系数，颜色，粗细，线条类型
imshow(image=image)
````
