使用`nvidia-smi`命令可以查看当前系统中安装的NVIDIA GPU的状态和信息，包括GPU型号、驱动版本、CUDA版本、显存使用情况等。注意，`nvidia-smi`命令显示的CUDA版本是驱动支持的最高CUDA版本，而不是当前安装的CUDA工具包版本

选择的 pytorch 版本：

- Python 3.10
- CUDA 12.8
- Torch 2.10

安装命令：

```bash
conda install torch==2.11.0 torchvision torchaudio
# or pip
pip install torch torchvision --index-url https://download.pytorch.org/whl/cu128 # PyTorch 的 CUDA 版本并不在标准的 PyPI 上，所以要指定源
```

使用如下Python代码来检查pytorch版本、CUDA是否可用以及CUDA版本

````python
import torch
print(torch.__version__) # check PyTorch version
print(torch.__file__) # check PyTorch installation path
print(torch.cuda.is_available()) # check if CUDA is available
print(torch.version.cuda) # check CUDA version
````

如果 `torch.cuda.is_available()` 返回 False，尝试以下步骤来解决问题：

1. 检查自己是不是下成CPU版torch了

````python
dir(torch) # 查看torch模块中的所有属性和方法
````
