# Data

在 PyTorch 中，Dataset 和 DataLoader 是数据加载与预处理的核心组件

Dataset 是一个抽象类，用于抽象表示一个数据集，负责存储样本及其标签

要使用 Dataset，我们需要创建一个继承自 Dataset 的子类，并实现以下两个方法：

- `__len__()`: 返回数据集的大小
- `__getitem__()`: 返回第 i 个数据及其对应的标签


```python
from torch.utils.data import Dataset # Utilities，是用于存放通用、复用性高的辅助函数的模块
import cv2
import os

class myDataset(Dataset):
    def __init__(self, root_dir, label_dir):
        self.root_dir = root_dir
        self.label_dir = label_dir
        self.path = os.path.join(root_dir, label_dir) # 将根目录和标签目录组合成完整的路径
        self.img_path = os.listdir(self.path) # 列出路径下的所有文件

    def __len__(self):
        return len(self.img_path) # 返回数据集中图像的数量

    def __getitem__(self, idx):
        img_name = self.img_path[idx]
        img_item_path = os.path.join(self.path, img_name) # 获取第 idx 个图像的完整路径
        img = cv2.imread(img_item_path) # 读取图像
        label = self.label_dir # 标签就是目录名
        return img, label
```

```python
myDataset = myDataset(root_dir="hymenoptera_data", label_dir="train")
print(myDataset.__len__()) # 输出数据集中图像的数量
```

DataLoader 将 Dataset 包装成一个可迭代对象，并提供批量采样、打乱顺序、多进程加载等实用功能

DataLoader 的主要参数包括：

- `dataset`: 需要加载的数据集（必须是 Dataset 的子类实例）
- `batch_size`: 每个批次的数据量
- `shuffle`: 是否在每个 epoch 开始时打乱数据
- `num_workers`: 使用多少个子进程并行加载数据
